```python
class Certificado:
    def __init__(self, numero_registro, identificacion, estudiante, programa, reporte, tipo_plantilla, fecha, estado):
        self.numero_registro = numero_registro
        self.identificacion = identificacion
        self.estudiante = estudiante
        self.programa = programa
        self.reporte = reporte
        self.tipo_plantilla = tipo_plantilla
        self.fecha = fecha
        self.estado = estado

    def mostrar_info(self):
        print(f"Registro #{self.numero_registro} | ID: {self.identificacion} | {self.estudiante}")
        print(f"   Programa: {self.programa}")
        print(f"   Reporte: {self.reporte} ({self.tipo_plantilla}) | Fecha: {self.fecha} | Estado: {self.estado}")


class NodoSimple:
    def __init__(self, certificado):
        self.certificado = certificado
        self.siguiente = None

class ColaListaEnlazada:
    def __init__(self):
        self.frente = None
        self.fin = None

    def encolar(self, certificado):
        nuevo = NodoSimple(certificado)
        if self.esta_vacia():
            self.frente = nuevo
            self.fin = nuevo
        else:
            self.fin.siguiente = nuevo
            self.fin = nuevo

    def desencolar(self):
        if self.esta_vacia():
            return None
        certificado = self.frente.certificado
        self.frente = self.frente.siguiente
        if self.frente is None:
            self.fin = None
        return certificado

    def esta_vacia(self):
        return self.frente is None


class NodoDoble:
    def __init__(self, certificado):
        self.certificado = certificado
        self.siguiente = None
        self.anterior = None

class ListaDobleCertificados:
    def __init__(self):
        self.cabeza = None
        self.cola = None

    def insertar(self, certificado):
        nuevo = NodoDoble(certificado)
        if self.cabeza is None:
            self.cabeza = nuevo
            self.cola = nuevo
        else:
            nuevo.anterior = self.cola
            self.cola.siguiente = nuevo
            self.cola = nuevo

    def buscar_por_registro(self, numero_registro):
        actual = self.cabeza
        while actual is not None:
            if actual.certificado.numero_registro == numero_registro:
                return actual.certificado
            actual = actual.siguiente
        return None

    def eliminar(self, numero_registro):
        actual = self.cabeza
        while actual is not None:
            if actual.certificado.numero_registro == numero_registro:
                if actual.anterior is not None:
                    actual.anterior.siguiente = actual.siguiente
                else:
                    self.cabeza = actual.siguiente

                if actual.siguiente is not None:
                    actual.siguiente.anterior = actual.anterior
                else:
                    self.cola = actual.anterior
                return True
            actual = actual.siguiente
        return False


class NodoArbol:
    def __init__(self, certificado):
        self.certificado = certificado
        self.izquierdo = None
        self.derecho = None

class ArbolCertificados:
    def __init__(self):
        self.raiz = None

    def insertar(self, certificado):
        if self.raiz is None:
            self.raiz = NodoArbol(certificado)
        else:
            self._insertar_recursivo(self.raiz, certificado)

    def _insertar_recursivo(self, nodo_actual, certificado):
        if certificado.numero_registro < nodo_actual.certificado.numero_registro:
            if nodo_actual.izquierdo is None:
                nodo_actual.izquierdo = NodoArbol(certificado)
            else:
                self._insertar_recursivo(nodo_actual.izquierdo, certificado)
        elif certificado.numero_registro > nodo_actual.certificado.numero_registro:
            if nodo_actual.derecho is None:
                nodo_actual.derecho = NodoArbol(certificado)
            else:
                self._insertar_recursivo(nodo_actual.derecho, certificado)

    def inorden(self, nodo_actual):
        if nodo_actual is not None:
            self.inorden(nodo_actual.izquierdo)
            nodo_actual.certificado.mostrar_info()
            print("-" * 50)
            self.inorden(nodo_actual.derecho)

    def calcular_altura(self, nodo_actual):
        if nodo_actual is None:
            return 0
        altura_izq = self.calcular_altura(nodo_actual.izquierdo)
        altura_der = self.calcular_altura(nodo_actual.derecho)
        return 1 + max(altura_izq, altura_der)


class GrafoProgramasSedes:
    def __init__(self):
        self.red = {}

    def agregar_nodo(self, nombre):
        if nombre not in self.red:
            self.red[nombre] = []

    def conectar(self, origen, destino):
        self.agregar_nodo(origen)
        self.agregar_nodo(destino)
        self.red[origen].append(destino)
        self.red[destino].append(origen)

    def mostrar_conexiones(self, lugar):
        if lugar in self.red:
            print(f"Relaciones de '{lugar}': {self.red[lugar]}")


if __name__ == "__main__":
    print("=== PLATAFORMA DE GESTIÓN Y REPORTES DE ESTUDIANTE ===")

    cola_procesamiento = ColaListaEnlazada()
    historial_reportes = ListaDobleCertificados()
    indice_arbol = ArbolCertificados()
    red_universidad = GrafoProgramasSedes()

    red_universidad.conectar("INGENIERIA DE SOFTWARE VIRTUAL", "SEDE MEDELLIN")

    c1 = Certificado(1005, "1090436483", "DHAYANNA ORTEGA", "INGENIERIA DE SOFTWARE VIRTUAL", "CERTIFICADO PROMEDIO", "NOTAS PREGRADO", "29-09-2026", "VALIDO")
    c2 = Certificado(1001, "1090436483", "DHAYANNA ORTEGA", "INGENIERIA DE SOFTWARE VIRTUAL", "CERTIFICADO DE ESTUDIO", "REGULAR", "30-09-2026", "VALIDO")

    cola_procesamiento.encolar(c1)
    cola_procesamiento.encolar(c2)

    print("\n1. Procesando solicitudes de la cola:")
    while not cola_procesamiento.esta_vacia():
        solicitud = cola_procesamiento.desencolar()
        print("  -> Solicitud Procesada:")
        solicitud.mostrar_info()
        print("-" * 50)

        historial_reportes.insertar(solicitud)
        indice_arbol.insertar(solicitud)

    print("\n2. Lista de Reportes Ordenados por Número de Registro (Inorden del Árbol):")
    indice_arbol.inorden(indice_arbol.raiz)

    print("\n3. Verificación de Grafo:")
    red_universidad.mostrar_conexiones("INGENIERIA DE SOFTWARE VIRTUAL")

    print("\n" + "=" * 60)
    print("      EXPERIMENTO COMPARATIVO DE ALTURAS (15 CLAVES)")
    print("=" * 60)

    claves_desordenadas = [1008, 1004, 1012, 1002, 1006, 1010, 1014, 1001, 1003, 1005, 1007, 1009, 1011, 1013, 1015]
    claves_ordenadas = [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010, 1011, 1012, 1013, 1014, 1015]

    arbol_exp_desordenado = ArbolCertificados()
    arbol_exp_ordenado = ArbolCertificados()

    for clv in claves_desordenadas:
        arbol_exp_desordenado.insertar(Certificado(clv, "1090436483", "PRUEBA", "ING SOFTWARE", "CERT", "PLANTILLA", "2026", "VALIDO"))

    for clv in claves_ordenadas:
        arbol_exp_ordenado.insertar(Certificado(clv, "1090436483", "PRUEBA", "ING SOFTWARE", "CERT", "PLANTILLA", "2026", "VALIDO"))

    h_desorden = arbol_exp_desordenado.calcular_altura(arbol_exp_desordenado.raiz)
    h_orden = arbol_exp_ordenado.calcular_altura(arbol_exp_ordenado.raiz)

    print(f"Cantidad de elementos insertados: 15")
    print(f"-> Altura con datos DESORDENADOS: {h_desorden}")
    print(f"-> Altura con datos ORDENADOS:    {h_orden}")
    print("=" * 60)
```
